Imports

```python
import serial
import matplotlib.pyplot as plt
import csv 
import pandas as pd
import plotly.express as px
```

Salva janela

```python
# Configuração da porta serial e do arquivo
porta = "COM3"
baud = 115200
arquivo = "dados.csv"

# Iniciando a leitura da porta serial
ser = serial.Serial(porta, baud)
ser.flush()
print("Abrindo Porta Serial")


amostra = 2000  # Número de amostras a capturar
linha = 0

# Abrindo o arquivo CSV para gravar as amostras
with open(arquivo, "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Valor1"])  # Cabeçalho do arquivo CSV

    while linha < amostra:
        data = ser.readline().decode("latin-1").strip()  # Lê e remove espaços e quebras de linha extras
        print(data)  # Mostra o valor lido no terminal

        # Divide os dados em dois valores separados por vírgula
        try:
            valor1 = float(data)
            writer.writerow([valor1])
            linha += 1
        except ValueError:
            print("Erro na leitura dos dados:", data)  # Tratamento de erro se os dados não estiverem no formato correto

# Fechando a conexão serial
ser.close()
print("\nFinalizado. Dados salvos em", arquivo)
```

Plot

```python
import csv
import matplotlib.pyplot as plt

# Listas para armazenar os valores de cada coluna
dados_valor1 = []
dados_valor2 = []

# Lendo os dados do arquivo CSV
with open("dados.csv") as file:
    reader = csv.reader(file, delimiter=',')
    next(reader)  # Pula o cabeçalho
    for row in reader:
        dados_valor1.append(float(row[0]))  # Adiciona o Valor1 à lista
        #dados_valor2.append(float(row[1]))  # Adiciona o Valor2 à lista

# Plotando os dados
plt.plot(dados_valor1, label="Valor1")  # Gráfico para Valor1
#plt.plot(dados_valor2, label="Valor2")  # Gráfico para Valor2
plt.xlabel("Amostra")
plt.ylabel("Valor")
plt.title("Dados da Porta Serial")
plt.legend()  # Mostra a legenda para diferenciar os valores
plt.show()
```
